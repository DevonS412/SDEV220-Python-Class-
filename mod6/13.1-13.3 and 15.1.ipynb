# 15.1 
Use multiprocessing to create three separate processes. Make each one wait a random number of seconds between zero and one, print the current time, and then exit.

```python
import multiprocessing
import random
import time

def worker(n):
    # Sleep for a random time between 0 and 1 seconds
    delay = random.random()
    time.sleep(delay)

    print(f"Process {n} waited {delay:.3f} seconds | Current time: {time.ctime()}")

if __name__ == "__main__":
    processes = []

    for i in range(3):
        p = multiprocessing.Process(target=worker, args=(i,))
        processes.append(p)
        p.start()

    # Optional: wait for all processes to finish
    for p in processes:
        p.join()
```
